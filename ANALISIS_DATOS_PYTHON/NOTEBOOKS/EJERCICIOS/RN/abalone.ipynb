```python
import keras
from keras.models import Sequential  
from keras import layers
from keras.layers import Dense  
import numpy
import pandas as pd  

import warnings
warnings.filterwarnings("ignore")
```

```python
name_columns = ['Sexo', 'Largura', 'Diámetro', 'Altura', 'Whole_weight', 'Shucked_weight', 'Viscera_weight', 'Shell_weight','Anillos']

dataset = pd.read_csv("C:/Users/Jose Manuel/Documents/VisualStudioCode_Workspace/Repo_CEDAPY/CEDAPY_25_26/ANALISIS_DATOS_PYTHON/NOTEBOOKS/EJERCICIOS/RN/abalone.data",
                        delimiter=",", header=None, names=name_columns)
display(dataset)
```

```python
map_target_class = {'M': 0, 'F': 1, 'I':3}
dataset['Sexo']= dataset['Sexo'].map(map_target_class)
dataset.head()
```

```python
X = dataset.iloc[:,0:8]  
Y = dataset.iloc[:,8]
```

```python
print(X.shape)
print(Y.shape)
```

```python
model2 = keras.Sequential([
    keras.layers.Input(shape=(8,)),
    keras.layers.Dense(12, activation='relu'),
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')    
])
```

```python
model2.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model2.fit(X, Y, epochs=150, batch_size=10)
```
