## Test Unitarios en Python Unittest y Pytest
La realización de un test unitario es el proceso por el que podemos comprobar el correcto funcionamiento de un fragmento de código o de la totalidad de una apliación. Cuando implementamos un conjunto de pruebas unitarias sobre nuestro código, hemos de comprobar que se cumplan los siguientes requisitos:

- **Automatizables**: Una vez implementado el conjunto de test unitarios, estos deben de tener la capacidad se ejecutarse y validadarse por si mismos, sin necesidad de una intervención manual del progrmador.
- **Completos**: Estos test deben cubrir la mayor cantidad posible de código desarrollado.
- **Reutilizables**: Deben ser ejecutables una y otra vez, sin que su resultado dependa de factores externos.
- **Independientes**: Cada test unitario debe ser independiente del resto, de forma que la ejecución de uno no afecte al resultado de otro.
- **Profesionales**: Deben ser desarrollados en conciencia y con la misma profesionalizadd con la que ha sido desarrollado el resto del código.

## Test unitarios para varias funciones   

Antes hemos visto como hacer un test unitario de una función, pero es posible que necesitemos hacer un test para varias funciones.    

A continuación vamos a suponer que tenemos un código en el que dentro nos encontramos con varias funciones que serían un conjunto de operadores matemáticos básicos.    
Estas funciones serían las siguientes:   

```python
def suma(x, y):
   return x + y
def resta(x, y):
   return x - y
def multiplicacion(x, y):
    return x * y
def division(x, y):
   if y == 0:
         raise ValueError("No se puede dividir por cero")  
      return x / y
```


Ahora crearemos un archivo llamado `test_operaciones.py` en el que se van a generar los test unitarios para cada una de estas funciones.   

```python

import pytest
from operaciones import suma, resta, multiplicacion, division

def test_suma():
   assert suma(3, 5) == 8
   assert suma(-1, 1) == 0
   assert suma(0, 0) == 0

def test_resta():
   assert resta(5, 3) == 2
   assert resta(-1, 1) == -2
   assert resta(0, 0) == 0

   assert multiplicacion(-1, 1) == -1

      assert division(0, 1) == 0
      with pytest.raises(ValueError):
         division(10, 0)
```

En este caso, se ha creado un test unitario para cada una de las funciones que se ha definido anteriormente.    
Cada test unitario comprueba diferentes casos para asegurarse de que las funciones funcionan correctamente.    

Ahora se puede ejecutar el comando `pytest test_operaciones.py` para comprobar que todos los test unitarios pasan correctamente.

Podemos usar el gestor de contexto `pytest.raises` para garantizar que una excepción se genere cuando esta se espera.   

Por ejemplo:

````python
import pytest
import ipytest
ipytest.autoconfig() # Configuración automática de ipytest. Solo en Jupyter Notebooks y una vez por notebook.

# Función a testear
def division(a, b):
    return a / b
 
# Test para la función division
def test_zero_division():
    with pytest.raises(ZeroDivisionError):
        division(10, 0)     
        

# ipytest.run() # Ejecuta todos los tests definidos en el notebook hasta el momento.

ipytest.run("-k test_zero_division") # Ejecuta un test específico; en este caso, el que verifica la excepción última
````

### Usando marcadores de test   

Si se quiere omitir un determinado test que no se desea ejecutar o se crea un test que se espera explicitamente que falle, podemos usar los marcadores.   

Un marcador de prueba es una etiqueta, concretamente, se usa el decorador `@pytest.mark` siguiendo la pauta del siguiente ejemplo:

````python
import pytest

def get_length(s):
    return len(s)

# Sintaxis de un marcador de pruebas
@pytest.mark.skip(reason="Saltando este test para demostración")
def test_get_length():
    assert get_length("hello") == 5
````

#### Tipos de marcadores     

```python
@pytest.mark.skip # marcador de salto, para omitir realizar un test.
@pytest.mark.skipif # marcador de salto condicional, sujeto a que se cumpla la condición de salto.
# Ejemplo de este último

import pytest

def get_length(s):
    return len(s)

# Sintaxis de un marcador de salto condicional
@pytest.mark.skipif('2 * 2 == 5') # la condición siempre será False y, por tanto, siempre se ejecuta el test.
def test_get_length():
    assert get_length("abc") == 3

@pytest.mark.xfail #marcador cuando se espera que un test falle.

# Ejemplo
import ipytest

def gen_sequence(n):
    return list(range(1, n+1))

# Usamos xfail
@pytest.mark.xfail
def test_gen_seq():
    assert gen_sequence(-1)

# En esta última prueba, se consigue que falle y el test nos indicará este hecho, tal y como esperábamos.
```

Más ejemplos de uso de marcadores:

````python
import datetime
import pytest


day_of_week = datetime.now().isoweekday()

def get_unique_values(lst):
    return list(set(lst))


condition_string = 'day_of_week == 6'
#Marcador condicional
@pytest.mark.skipif(condition_string)
def test_function():
	# Lineas de testeo
    assert get_unique_values([1,2,3]) == [1,2,3]
    assert get_unique_values([1,2,3,1]) == [1,2,3]
````

````python
def multiple_of_two(num):
    if num == 0:
        raise(ValueError)
    return num % 2 == 0

# Marcador que indica que el test fallará
@pytest.mark.xfail
def test_fails():
    # Test que fallará intencionadamente
    assert multiple_of_two(3) is True
````

## Fixtures, los contextos específicos de test   

### ¿Qué son las fixtures?

En testing, una ***fixture*** proporciona un **contexto definido, fiable y consistente** para los tests (por ejemplo, un dataset, un fichero temporal, una conexión, una configuración). En pytest, ese contexto se implementa como **funciones** decoradas con `@pytest.fixture`, y los tests acceden a ellas como **argumentos**.

### Ventajas frente a setup/teardown “clásico” (xUnit)
Las fixtures mejoran el enfoque típico `setUp/tearDown` porque:
- Son **explícitas** (se activan porque las declaras en el test).
- Son **modulares** (una fixture puede depender de otras).
- Escalan bien (reutilización, scopes, parametrización).
- Facilitan teardown seguro incluso con muchas dependencias.

---   


### Primer contacto: “pedir” fixtures desde un test

#### Ejemplo mínimo paso a paso

**Estructura:**

mi_proyecto/    
    tests/   
        test_fruta.py

````python
# Contenido del fichero tests/test_fruta.py

import pytest

class Fruit:
    def __init__(self, name):
        self.name = name

@pytest.fixture
def my_fruit():
    return Fruit("apple")

def test_name_is_apple(my_fruit):
    assert my_fruit.name == "apple"
````

Qué ocurre aquí (paso a paso):

1. Pytest detecta `test_name_is_apple` y ve que tiene un parámetro: `my_fruit`.
2. Busca una ***fixture*** con el mismo nombre: `my_fruit`.
3. Ejecuta la ***fixture*** `my_fruit()`, recoge su *return* y lo inyecta como argumento.
4. Ejecuta el test con ese objeto.

### Fixtures que dependen de otras fixtures (composición)

Una de las grandes ventajas es que las **fixtures** pueden pedir otras **fixtures**, igual que un test.

````python
import pytest

@pytest.fixture
def first_entry():
    return "a"

@pytest.fixture
def order(first_entry):
    return [first_entry]

def test_append(order):
    order.append("b")
    assert order == ["a", "b"]
````

En el ejemplo anterior:   
- `order` depende de `first_entry`.
- Pytest resuelve el “grafo” de dependencias y ejecuta en orden correcto

### Reutilización y caché de fixtures (dentro de un test)
#### Reutilización entre tests

Dos tests pueden pedir la misma **fixture** y pytest ejecutará la **fixture** una vez por test (por defecto).

#### Caché dentro del mismo test

Si una **fixture** se pide varias veces (directa o indirectamente) no se re-ejecuta dentro del mismo test: su valor se cachea para ese test.   

Ejemplo:

````python
import pytest

@pytest.fixture
def order():
    return []

@pytest.fixture
def first_entry():
    return "a"

@pytest.fixture
def append_first(order, first_entry):
    order.append(first_entry)

def test_cached(order, first_entry, append_first):
    # 'order' es el mismo objeto que modificó append_first
    assert order == [first_entry]
````

### Autouse fixtures (se ejecutan sin pedirlas)

A veces es necesario que siempre se ejecute una preparación (por ejemplo, reset de estado). Para eso existe `autouse=True`.

````python
import pytest

@pytest.fixture
def order():
    return []

@pytest.fixture(autouse=True) # Auto-use fixture.
def add_default(order):
    order.append("default")

def test_one(order):
    assert order == ["default"]

def test_two(order):
    order.append("x")
    assert order == ["default", "x"]
````

*Idea clave*: los tests no mencionan `add_default`, pero les afecta porque se ejecuta automáticamente en su ámbito.

### Scope: controlar cuántas veces se crea una **fixture**

Por defecto, el scope es tipo `function` (una vez por test). Pero si se considera costoso crear/usar un recurso concreto (red, BBDD, contenedores…), es posible ampliar **scope**.

#### Scopes más comunes

- ***function***: por test (por defecto)
- ***class***: por clase de tests
- ***module***: por fichero test_*.py
- ***package***: por paquete
- ***session***: una vez en toda la sesión de pytest    

Ejemplo con `scope="module"`:

*Regla práctica*: un fixture de scope “amplio” puede usar fixtures de scope igual o más amplio; al revés no suele tener sentido (y pytest lo evita).

````python
import pytest

@pytest.fixture(scope="module")
def expensive_resource():
    # Simula una creación "costosa" de recurso
    return {"conn": "created_once_per_module"}

def test_a(expensive_resource):
    assert expensive_resource["conn"] == "created_once_per_module"

def test_b(expensive_resource):
    assert "conn" in expensive_resource
````

### Teardown / Cleanup: fixtures con `yield` (recomendado)

Cuando el test termina, normalmente es interesante limpiar (cerrar conexiones, borrar ficheros, etc.). Pytest recomienda fixtures con `yield`: lo que va antes del `yield` prepara; lo que va después limpia.

#### Ejemplo simple paso a paso:

````python
import pytest

@pytest.fixture
def temp_list():
    data = []
    # setup
    yield data
    # teardown (limpieza)
    data.clear()

def test_use(temp_list):
    temp_list.append(1)
    assert temp_list == [1]
````

En el código anterior, `pytest` realiza lo siguiente:    

1. Ejecuta temp_list hasta el yield y entrega data al test.

2. Ejecuta el test.

3. Vuelve a temp_list y ejecuta el código tras el yield (cleanup).

### `usefixtures`: usar fixtures aunque no necesites su valor

Hay situaciones en las que puede ser necesario disponer del efecto de la fixture (p. ej., cambiar a un directorio vacío), pero no es conveniente o necesario disponer de un objeto. En ese caso, es recomendable usar `@pytest.mark.usefixtures("nombre")`.    

Ejemplo:

````python
import os
import pytest
import tempfile

# normalmente viviría en conftest.py
@pytest.fixture
def cleandir():
    with tempfile.TemporaryDirectory() as newpath:
        old = os.getcwd()
        os.chdir(newpath)
        yield
        os.chdir(old)

@pytest.mark.usefixtures("cleandir")
class TestDirectory:
    def test_starts_empty(self):
        assert os.listdir(os.getcwd()) == []
````

### Compartir fixtures con `conftest.py`

`conftest.py` permite definir fixtures para un directorio completo; pytest las descubre automáticamente, y pueden existir `conftest.py` anidados.

Ejemplo de estructura:   

tests/   
conftest.py    
test_a.py   
subpackage/    
conftest.py   
test_b.py    


*Idea clave*: un test puede **“buscar hacia arriba”** fixtures en`conftest.py` padres, pero no “bajar” a otros módulos hermanos.

### Sobreescritura de fixtures (**override**) por nivel

En proyectos grandes, es posible redefinir una fixture con el mismo nombre:

* En un subdirectorio con su `conftest.py`
* En un módulo de test concreto
* Incluso sobrescribir por parametrización directa del test

Ejemplo conceptual (módulo):

````python
# conftest.py (global)
import pytest

@pytest.fixture
def username():
    return "username"
````

````python
# test_something.py (override local)
import pytest

@pytest.fixture
def username(username):
    return "overridden-" + username

def test_username(username):
    assert username == "overridden-username"
````

### Parametrizar fixtures con params y request.param

Una fixture puede generar múltiples variantes con `params=[...]`. Pytest ejecutará los tests dependientes una vez por cada parámetro, y el valor actual se accede con `request.param`.

````python
import pytest

@pytest.fixture(params=[0, 1, 2])
def number(request):
    return request.param

def test_is_int(number):
    assert isinstance(number, int)
````

#### Parametrización + marcas por caso

Es posible usar `pytest.param(..., marks=...)` para marcar un caso (por ejemplo `skip`)

### Fixtures “factoría” (`factory fixtures`)

Cuando en un mismo test es necesario crear muchos objetos parecidos, una fixture puede devolver una función “creadora”.   

A continuación se muestra un ejemplo de factoría simple:

````python
import pytest

@pytest.fixture
def make_user():
    def _make_user(name):
        return {"name": name}
    return _make_user

def test_multiple(make_user):
    u1 = make_user("Ana")
    u2 = make_user("Luis")
    assert u1["name"] == "Ana"
    assert u2["name"] == "Luis"
````

El siguiente ejemplo presenta una factoría con *cleanup" (guardado de lo creado):

````python
import pytest

@pytest.fixture
def make_record():
    created = []

    def _make_record(value):
        rec = {"value": value, "alive": True}
        created.append(rec)
        return rec

    yield _make_record

    # cleanup
    for rec in created:
        rec["alive"] = False

def test_records(make_record):
    r1 = make_record(10)
    r2 = make_record(20)
    assert r1["alive"] and r2["alive"]
````

### Fixtures incorporadas (built-in) que conviene conocer

Pytest trae fixtures listas para usar, por ejemplo:

* tmp_path: directorio temporal único por test
* monkeypatch: parcheo temporal de funciones/atributos/env
* capsys / capfd: captura de salida estándar
* caplog: captura/control de logging
* request: información del test/fixture en ejecución

***Consejo***: estas fixtures reducen muchísimo la necesidad de “inventar” infraestructura propia.

## Test Unitarios con Unittest   

Otra de las librerías que podemos utilizar para realizar test unitarios en Python es **Unittest**. Esta librería viene incluida en la biblioteca estándar de Python, por lo que no es necesario instalar nada adicional para utilizarla.

**Unittest** nos va a ofrecer tres tipos de salidas:

- **OK**: Indica que la prueba ha sido ejecutada correctamente y no se ha encontrado ningún error.
- **FAIL**: Indica que la prueba ha fallado debido a una aserción y que es necesario revisar el código para corregir el error. Además sería interesante incluir una excepción que nos indique el motivo del fallo (try/except).
- **ERROR**: Indica que se ha producido un error durante la ejecución y no ha pasado la prueba.    

Debemos comprobar que tenemos la librería *unittest2* instalada. Si no la tenemos, podemos instalarla utilizando el siguiente comando:

````python
!pip install unittest2
````

Vamos a ver un ejemplo de cómo utilizar Unittest para crear un test unitario para una función que suma dos números enteros.

```python
import unittest

def suma(x, y):
   return x + y

class TestSuma(unittest.TestCase):

   def test_suma(self):
      self.assertEqual(suma(3, 5), 8)
      self.assertEqual(suma(-1, 1), 0)
      self.assertEqual(suma(0, 0), 0)

if __name__ == '__main__':
   unittest.main()

```

Una vez que lo ejecutemos veremos en la terminal que todos los test unitarios han pasado correctamente.    
Cómo se puede observar, la gran diferencia entre las herramientas que hemos visto para realizar test unitarios en Python es la forma en la que se escriben los test unitarios.    

Ambas herramientas son muy útiles y permiten realizar test unitarios de una forma sencilla y cómoda. Las diferencias más destacables son:

- Implementar una clase que herede de la clase `unittest.TestCase`.
- Necesitamos utilizar la palabra clave reservada **assertEqual** para comprobar que el resultado de nuestra función implementada devuelve el resultado correcto.

## Limitaciones en los test unitarios   

Hay pruebas que serían complejas de hacer, como por ejemplo:

- Pruebas que dependan de la interfaz gráfica de usuario (GUI).
- Pruebas que requieran interacción con hardware específico.
- Pruebas que dependan de servicios externos o APIs.
- Pruebas que involucren condiciones de carrera o concurrencia.
- Pruebas que requieran un entorno específico o configuración compleja.
- Pruebas que dependan de datos dinámicos o en tiempo real.
- Pruebas que involucren algoritmos complejos o cálculos matemáticos avanzados.   

Por ejemplo:    
    si se requiere hacer una prueba a un generador de números aleatorios y si, obviamente, no salen los mismos números, no es posible simular cada uno de los casos posibles con cada uno de los números que tenemos.    
    En esa situación, lo más conveniente sería probar que ocurre cuando se acierta o no se acierta un número, pero no probar con todos los números posibles.   

## Ejemplo de uso con **Unitest**   

Aunque el uso de `assert()` puede ser suficiente para algunos tests, a veces se nos queda corto y se necesitan librerías como *unittest*, que ofrecen alguna que otra funcionalidad que hará la vida más fácil. 
Veamos un ejemplo. Tenemos la función `calcula_media`, que es la que queremos testear.   

```python
# funciones.py
def calcula_media(*args):
    return(sum(*args)/len(*args))
```

Es posible usar *unittest* para crear varios tests que verifiquen que la función funciona correctamente. Aunque la estructura de un conjunto de tests se puede complicar más, la estructura será siempre muy similar a la siguiente:

- Se crea una clase `Test<NombreDeLoQueSePrueba>` que hereda de `unittest.TestCase`.
- Se definen varios tests como métodos de la clase, usando `test_<NombreDelTest>` para nombrarlos.
- En cada test se ejecutan las comprobaciones necesarias, usando `assertEqual` en lugar de `assert`, pero su comportamiento es totalmente análogo.   

```python
# tests.py
from funciones import calcula_media
import unittest

class TestCalculaMedia(unittest.TestCase):
    def test_1(self):
        resultado = calcula_media([10, 10, 10])
        self.assertEqual(resultado, 10)

    def test_2(self):
        resultado = calcula_media([5, 3, 4])
        self.assertEqual(resultado, 4)

if __name__ == '__main__':
    unittest.main() 
```    

Si se ejecuta el código anterior, se obtiene el siguiente resultado.    
Esta es una de las ventajas de unittest, ya que muestra información sobre los tests ejecutados, el tiempo que ha tardado y los resultados.   
```bash
Ran 2 tests in 0.006s

OK   
```
Por otro lado, usando `-v` es posible obtener más información sobre cada test ejecutado con su resultado individualmente.    
Si se tiene gran cantidad de tests suele ser recomendable usarla, ya que será más fácil localizar los tests que han fallado.   

```bash
$ python -m unittest -v tests

test_1 (tests.TestCalculaMedia) ... ok
test_2 (tests.TestCalculaMedia) ... ok

----------------------------------------------------------------------
Ran 2 tests in 0.000s

OK
```
Por último, si se tienene varios archivos de test, es posible usar `discover`, para indicar a Python que busque en la carpeta todos los tests y los ejecute.   
```bash
$ python -m unittest discover -v
```

### Usando `setUp` y `tearDown`   

Otra de las ventajas de usar **unittest**, es que ofrece la posibilidad de definir funciones comunes que son ejecutadas *antes* y *después* de cada test. Estos métodos son `setUp()` y `tearDown()`.

```python
import unittest
class TestEjemplos(unittest.TestCase):
    def setUp(self):
        print("Entra setUp")
    def tearDown(self):
        print("Entra tearDown")

    def test_1(self):
        print("Test: test_1")
    def test_2(self):
        print("Test: test_2")   
```
Se puede ver que hace una especie de sandwich con cada test, metiéndolo entre `setUp` y `tearDown`.    

Nótese que si ambas funciones realizan siempre lo mismo, es posible usar a modo de TestSuite con una función común para todos los tests.
```bash
$ python -m unittest -v tests

test_1 (tests.TestEjemplos) ... Entra setUp
Test: test_1
Entra tearDown
ok
test_2 (tests.TestEjemplos) ... Entra setUp
Test: test_2
Entra tearDown
ok

----------------------------------------------------------------------
Ran 2 tests in 0.000s

OK   
```

## Conclusión    

Las pruebas unitarias son una parte fundamental del desarrollo de software y nos permiten asegurar la calidad y funcionalidad de nuestro código.    

En Python, tenemos varias herramientas disponibles para implementar pruebas unitarias, siendo **Pytest** y **Unittest** dos de las más populares.    

Ambas herramientas nos ofrecen diferentes enfoques y características para crear y ejecutar pruebas unitarias de manera efectiva. Al elegir entre **Pytest** y **Unittest**, es importante considerar las necesidades específicas de nuestro proyecto y el equipo de desarrollo.    

Independientemente de la herramienta que elijamos, l*o más importante es adoptar una cultura de pruebas unitarias en nuestro proceso de desarrollo* para garantizar la calidad y confiabilidad de nuestro código.
